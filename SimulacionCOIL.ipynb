Hay que reproducir la fórmula de rabi (nuestra ecuación 18 del informe)
$$P_{1\to 2}(t) = \frac{4|H_{21}|^2}{(H_{11}-H_{22})^2 + 4|H_{21}|^2} \sin^2\left( \frac{\Omega_R t}{2} \right) = \frac{4|H_{21}|^2}{\hbar^2 \Omega_R^2} \sin^2\left( \frac{\Omega_R t}{2} \right)$$

Primero hay que definir los parámetros $H_11 $, etc del hamiltoniano:
$$\mathbf{H} =
  \begin{pmatrix} H_{11} & H_{12} \\ H_{21} & H_{22} \end{pmatrix} $$

```python
#definir parámetros:
H11=0
H22=0
H21=1
hbarra= 1.054e-34
```

También hay que calcular estas cosas:
$c = \frac{1}{2}\Delta E = \frac{1}{2}\sqrt{(H_{11}-H_{22})^2 + 4|H_{21}|^2}$

$\textit{frecuencia generalizada de Rabi}$: $\Omega_R \equiv \frac{2c}{\hbar} = \frac{\Delta E}{\hbar} $

$$\Omega_R \equiv \frac{2c}{\hbar} = \frac{\Delta E}{\hbar} = \frac{1}{\hbar}\sqrt{(H_{11}-H_{22})^2 + 4|H_{21}|^2} $$

```python
#definiendo la frecuencia generalizada de rabi:

frabi = (1/hbarra)* np.sqrt((H11-H22)**2 + 4*(np.abs(H21)**2))
frabi
```

entonces definiendo la función de rabi:

```python
def probabilidadRABI(t):
    deltaE = hbarra*frabi
    A = (4*np.abs(H21)**2)/ (deltaE**2)
    return A*np.sin(frabi*t /2)**2
```
